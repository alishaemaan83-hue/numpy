```python
import numpy as np
import time

# Python list approach
lst = list(range(1_000_000))
t0 = time.time()
result = [x * 2 for x in lst]
print(f"List:  {(time.time()-t0)*1000:.1f} ms")

# NumPy approach
arr = np.arange(1_000_000)
t0 = time.time()
result = arr * 2
print(f"NumPy: {(time.time()-t0)*1000:.1f} ms")
```

```python
import numpy as np

# 1D array
a = np.array([1, 2, 3, 4, 5])
print(a)          # [1 2 3 4 5]
print(a.dtype)    # int64
print(a.shape)    # (5,)
print(a.ndim)     # 1

# 2D array (matrix)
b = np.array([[1, 2, 3],
              [4, 5, 6]])
print(b.shape)    # (2, 3)
```
